```python
import sys
from rich import print
from time import sleep

def hide_cursor():
    sys.stdout.write("\033[?25l")
    sys.stdout.flush()

def show_cursor():
    sys.stdout.write("\033[?25h")
    sys.stdout.flush()

def printLyric():
    lines = [
        "نذرٍ علي لا أحرسك و أفديك و أغير",
        "ديني وعرضي فيك و أغلى حلالي",
        "و اللي يأثر فيك بـ أيات تأثير",
        "أبرأ منه لو كان عمي و خالي",
        "و أن مت و أخذتني صروف المقادير",
        "و الله ! لا أوصي فيك حتى عيالي"
    ]

    delays = [3.5, 4, 3.5, 2, 2, 2]

    hide_cursor()

    try:
        for i, line in enumerate(lines):
            print(f"[green]{line}[/green]")
            sleep(delays[i])
    finally:
        show_cursor()

printLyric()
```
